```python
import os

def get_modality_paths(patient_folder_path):
  """
  Generates a dictionary of full file paths for MRI modalities and segmentation masks
  for a given patient folder.

  Args:
    patient_folder_path (str): The path to the patient's data folder.

  Returns:
    dict: A dictionary where keys are modality names (t1c, t1n, t2f, t2w, seg)
          and values are their corresponding full file paths.
  """
  modalities = ['t1c', 't1n', 't2f', 't2w', 'seg']
  modality_paths = {}

  # Extract patient_id from the patient_folder_path
  patient_id = os.path.basename(patient_folder_path)

  for modality in modalities:
    file_name = f"{patient_id}-{modality}.nii.gz"
    full_path = os.path.join(patient_folder_path, file_name)
    modality_paths[modality] = full_path

  return modality_paths

print("The 'get_modality_paths' function has been defined.")
```

```python
import nibabel as nib
import matplotlib.pyplot as plt

def visualize_modality_brain_tumors(sample_patient_folder):
  modality_paths = get_modality_paths(sample_patient_folder) # This variable is already defined and available in the kernel

  modalities_to_display = ['t1c', 't1n', 't2f', 't2w', 'seg']
  num_modalities = len(modalities_to_display)

  fig, axes = plt.subplots(1, num_modalities, figsize=(4 * num_modalities, 5))

  if num_modalities == 1:
      axes = [axes] # Ensure axes is iterable even for a single subplot

  patient_name = os.path.basename(sample_patient_folder)
  fig.suptitle(f'MRI Modalities and Segmentation for Patient: {patient_name}', fontsize=16)

  for i, modality in enumerate(modalities_to_display):
      path = modality_paths.get(modality)
      if path and os.path.exists(path):
          try:
              img = nib.load(path)
              data = img.get_fdata()
              print(data.shape)

              # Get middle slice for 3D images
              if data.ndim == 3:
                  slice_to_display = data[:, :, data.shape[2] // 2]
              elif data.ndim == 2:
                  slice_to_display = data # Already a 2D image
              else:
                  print(f"Warning: {modality} image has unexpected dimensions: {data.ndim}. Skipping.")
                  axes[i].set_title(f'{modality} (Error)')
                  axes[i].axis('off')
                  continue

              # Use different colormap for segmentation for better visual distinction
              cmap = 'gray' if modality != 'seg' else 'hot'

              axes[i].imshow(slice_to_display, cmap=cmap)
              axes[i].set_title(modality.upper())
              axes[i].axis('off')

          except Exception as e:
              print(f"Error loading or processing {modality} for patient {patient_name}: {e}")
              axes[i].set_title(f'{modality} (Error)')
              axes[i].axis('off')
      else:
          print(f"Path not found for {modality}: {path}")
          axes[i].set_title(f'{modality} (Missing)')
          axes[i].axis('off')

  plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # Adjust layout to prevent title overlap
  plt.show()


# Reuse the sample_patient_folder and modality_paths from the previous step
sample_patient_folder = folders[5] # This variable is already defined and available in the kernel
visualize_modality_brain_tumors(sample_patient_folder)
```

```python
import nibabel as nib
import matplotlib.pyplot as plt
import os

def visualize_3_planes_modality_brain_tumors(patient_folder_path):
  """
  Visualizes all MRI modalities (t1c, t1n, t2f, t2w, seg) for a given patient
  in axial, sagittal, and coronal views.

  Args:
    patient_folder_path (str): The path to the patient's data folder.
  """
  modality_paths = get_modality_paths(patient_folder_path)

  modalities_to_display = ['t1c', 't1n', 't2f', 't2w', 'seg']
  planes = ['Axial', 'Sagittal', 'Coronal']
  num_modalities = len(modalities_to_display)
  num_planes = len(planes)

  fig, axes = plt.subplots(num_modalities, num_planes, figsize=(num_planes * 4, num_modalities * 4))

  patient_id = os.path.basename(patient_folder_path)
  fig.suptitle(f'MRI Modalities and Segmentation for Patient: {patient_id}', fontsize=16)

  for i, modality in enumerate(modalities_to_display):
    path = modality_paths.get(modality)
    if path and os.path.exists(path):
      try:
        img = nib.load(path)
        data = img.get_fdata()

        # Define colormap based on modality
        cmap = 'gray' if modality != 'seg' else 'hot'

        # Middle slices for axial, sagittal, coronal views
        axial_slice = data[:, :, data.shape[2] // 2] # x, y, z -> z-axis (axial)
        sagittal_slice = data[data.shape[0] // 2, :, :] # x, y, z -> x-axis (sagittal)
        coronal_slice = data[:, data.shape[1] // 2, :] # x, y, z -> y-axis (coronal)

        slices_to_plot = [axial_slice, sagittal_slice, coronal_slice]

        for j, (slice_data, plane_name) in enumerate(zip(slices_to_plot, planes)):
          ax = axes[i, j]
          ax.imshow(slice_data, cmap=cmap)
          ax.set_title(f'{modality.upper()} - {plane_name}')
          ax.axis('off')

      except Exception as e:
        for j in range(num_planes):
          ax = axes[i, j]
          ax.set_title(f'{modality.upper()} (Error)')
          ax.text(0.5, 0.5, f'Error: {e}', horizontalalignment='center', verticalalignment='center', transform=ax.transAxes, color='red')
          ax.axis('off')
        print(f"Error loading or processing {modality} for patient {patient_id}: {e}")
    else:
      for j in range(num_planes):
        ax = axes[i, j]
        ax.set_title(f'{modality.upper()} (Missing)')
        ax.text(0.5, 0.5, 'File Not Found', horizontalalignment='center', verticalalignment='center', transform=ax.transAxes, color='red')
        ax.axis('off')
      print(f"Path not found for {modality} for patient {patient_id}: {path}")

  plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # Adjust layout to prevent title overlap
  plt.show()

print("The 'visualize_3_planes_modality_brain_tumors' function has been defined.")
```

### Common Functions

```python
import cupy as cp # Changed from numpy as np
from tqdm import tqdm
import nibabel as nib # Added for self-containment
import os # Added for self-containment (basename)
import glob # Added for self-containment
import matplotlib.pyplot as plt


def rm_zero_values(img_data_np, range_start: int, range_end: int):
  """
  Filter out all-zero slices using CuPy.
  Args:
    img_data_np (numpy.ndarray): The input image data as a NumPy array.
  Returns:
    cupy.ndarray: A CuPy array containing only non-zero axial slices.
  """
  img_data_cp = cp.asarray(img_data_np) # Convert to CuPy array

  filtered_slices_cp = []

  # Iterate through the depth (z-axis) to 155
  # for index_slice in range(img_data_cp.shape[2]):
  for index_slice in range(range_start, range_end):
      # Extract the axial slice
      axial_slice_cp = img_data_cp[:, :, index_slice]

      # Keep only slices that are NOT all zeros
      if not cp.all(axial_slice_cp == 0):
          filtered_slices_cp.append(axial_slice_cp)

  # Convert list of CuPy arrays back to a single CuPy array
  if filtered_slices_cp:
      return cp.stack(filtered_slices_cp, axis=2)
  else:
      # If no non-zero slices, return an empty array with appropriate dimensions
      # Assuming 2D slices for H, W. Depth will be 0.
      return cp.empty((img_data_cp.shape[0], img_data_cp.shape[1], 0), dtype=img_data_cp.dtype)


def validated_check_zeros(data, slice_index):
  img_data_np = nib.load(data).get_fdata() # nibabel returns numpy array
  img_data_cp = cp.asarray(img_data_np) # Convert to CuPy array
  axial_slice_cp = img_data_cp[:, :, slice_index]
  return cp.all(axial_slice_cp == 0)

def display_axial_slice(file_path, slice_index):
  """
  Loads a NIfTI file and displays a specific axial slice.

  Args:
    file_path (str): The full path to the NIfTI file.
    slice_index (int): The index of the axial slice to display.
  """
  try:
    img = nib.load(file_path)
    data_np = img.get_fdata() # Get data as a NumPy array

    if data_np.ndim < 3:
      print(f"Error: Image at {file_path} is not 3D. Cannot display axial slice.")
      return

    if not (0 <= slice_index < data_np.shape[2]):
      print(f"Error: Slice index {slice_index} is out of bounds for image with depth {data_np.shape[2]}.")
      return

    axial_slice = data_np[:, :, slice_index]

    plt.figure(figsize=(6, 6))
    plt.imshow(axial_slice, cmap='gray')
    plt.title(f'Axial Slice {slice_index} from {os.path.basename(file_path)}')
    plt.axis('off')
    plt.show()

  except FileNotFoundError:
    print(f"Error: File not found at {file_path}")
  except Exception as e:
    print(f"An error occurred: {e}")
```

```python
import torch
from torch.utils.data import Dataset, DataLoader

import os
import nibabel as nib
import numpy as np
from monai.transforms import HistogramNormalize, ScaleIntensityRange

class MRIDataset(Dataset):
  # ['t1c', 't1n', 't2f', 't2w', 'seg']
  def __init__(self, root_dir, input_modalities=['t1n', 't2f', 't2w'], target_modality='t1c'):
    self.root_dir = root_dir
    self.input_modalities = input_modalities
    self.target_modality = target_modality

    self.slice_start = 15
    # Adjusted slice_end to ensure the depth (number of slices) is divisible by 16
    # (111 - 15 = 96, and 96 is divisible by 16 for UNet with 4 strides of 2)
    self.slice_end = 111

    # Initialize transforms
    self.hist_normalize = HistogramNormalize()
    # MinMax scaling will be applied per image after histogram normalization
    # The range [0, 1] is a common practice for neural networks
    self.scale_intensity = ScaleIntensityRange(a_min=0.0, a_max=255.0, b_min=0.0, b_max=1.0) # Assumed range after HistogramNormalize

  def __len__(self):
    return len(self.root_dir)

  def __getitem__(self, idx):
    patient_folder_path = self.root_dir[idx]
    modality_paths = get_modality_paths(patient_folder_path) # Reuse the helper function

    # Collect input modalities
    processed_inputs = []

    for modality in self.input_modalities:
      path = modality_paths.get(modality)
      if not path or not os.path.exists(path):
        # For a robust pipeline, you might want to handle this more gracefully,
        # e.g., by skipping the patient or returning an error.
        raise FileNotFoundError(f"Required input modality {modality} not found for patient {os.path.basename(patient_folder_path)} at {path}")

      img = nib.load(path).get_fdata() # Load as NumPy array
      img = img[:,:, self.slice_start:self.slice_end] # Axial views as the fields
      img_tensor = torch.from_numpy(img).float() # Convert to PyTorch tensor

      # Apply Histogram Normalization
      img_normalized = self.hist_normalize(img_tensor)

      # Apply MinMax Scaling
      img_scaled = self.scale_intensity(img_normalized)

      # Add a channel dimension for concatenation, assuming (H, W, D) -> (1, H, W, D)
      processed_inputs.append(img_scaled.unsqueeze(0))

    # Concatenate input modalities along the channel dimension
    # Resulting shape will be (num_input_modalities, H, W, D)
    input_tensor_concatenated = torch.cat(processed_inputs, dim=0)

    # Load and process target modality (t1c)
    target_path = modality_paths.get(self.target_modality)
    if not target_path or not os.path.exists(target_path):
      raise FileNotFoundError(f"Target modality {self.target_modality} not found for patient {os.path.basename(patient_folder_path)} at {target_path}")

    target_img = nib.load(target_path).get_fdata()
    target_img = target_img[:, :, self.slice_start:self.slice_end] #Axial views
    target_tensor = torch.from_numpy(target_img).float()

    # Apply Histogram Normalization to target
    target_normalized = self.hist_normalize(target_tensor)

    # Apply MinMax Scaling to target
    target_scaled = self.scale_intensity(target_normalized)

    # Add a channel dimension for the target: (H, W, D) -> (1, H, W, D)
    target_tensor_final = target_scaled.unsqueeze(0)

    return {'inputs': input_tensor_concatenated, 'target': target_tensor_final}
```

```python
import torch
import torch.nn as nn
from monai.networks.nets import UNet
from monai.networks.layers import Act, Norm


# --- Generator Architecture (U-Net) ---
# The Generator is a U-Net that takes multiple input modalities
# and generates the target modality.
class Generator(nn.Module):
  def __init__(self, in_channels, out_channels):
    super().__init__()
    self.unet = UNet(
        spatial_dims=3,  # 3D medical images
        in_channels=in_channels,  # Number of input modalities (e.g., 3 for t1n, t2f, t2w)
        out_channels=out_channels, # Number of output modalities (e.g., 1 for t1c)
        channels=(16, 32, 64, 128, 256), # Number of feature channels at each layer
        strides=(2, 2, 2, 2), # Stride for convolutional layers (downsampling)
        num_res_units=2, # Number of residual units in each block
        norm=Norm.INSTANCE, # Instance normalization
        act=Act.LEAKYRELU, # Leaky ReLU activation
    )

  def forward(self, x):
    return self.unet(x)

# --- Discriminator Architecture (PatchGAN) ---
# The Discriminator is a convolutional network that classifies whether
# an image pair (input, target) is 'real' or 'fake'.
# It outputs a 'patch' of classifications rather than a single value.
class Discriminator(nn.Module):
  def __init__(self, in_channels, out_channels):
    super().__init__()
    # The discriminator receives both the input image and the generated/real image.
    # So its input channels are in_channels (input modalities) + out_channels (generated/real target)
    self.model = nn.Sequential(
        # Initial convolutional layer
        nn.Conv3d(in_channels + out_channels, 64, kernel_size=4, stride=2, padding=1),
        nn.LeakyReLU(0.2, inplace=True),

        # Hidden layers
        nn.Conv3d(64, 128, kernel_size=4, stride=2, padding=1, bias=False),
        nn.InstanceNorm3d(128),
        nn.LeakyReLU(0.2, inplace=True),

        nn.Conv3d(128, 256, kernel_size=4, stride=2, padding=1, bias=False),
        nn.InstanceNorm3d(256),
        nn.LeakyReLU(0.2, inplace=True),

        nn.Conv3d(256, 512, kernel_size=4, stride=1, padding=1, bias=False), # stride=1, no downsampling
        nn.InstanceNorm3d(512),
        nn.LeakyReLU(0.2, inplace=True),

        # Output layer - 1 channel output representing real/fake for each patch
        nn.Conv3d(512, 1, kernel_size=4, stride=1, padding=1)
    )

  def forward(self, input_image, target_image):
    # Concatenate input and target images along the channel dimension
    x = torch.cat([input_image, target_image], 1)
    return self.model(x)

# --- Weight Initialization ---
def weights_init(m):
  classname = m.__class__.__name__
  if classname.find('Conv') != -1:
    nn.init.normal_(m.weight.data, 0.0, 0.02)
  elif classname.find('BatchNorm') != -1:
    nn.init.normal_(m.weight.data, 1.0, 0.02)
    nn.init.constant_(m.bias.data, 0)

print("Generator and Discriminator architectures defined.")
print("Weight initialization function defined.")
```

```python
import os
import nibabel as nib
import numpy as np
import torch
from torch.utils.data import Dataset
from monai.transforms import HistogramNormalize, ScaleIntensityRange
import random
import glob
from torch.utils.data import DataLoader

class MRI3DVoxelDataset(Dataset):
    def __init__(self, patient_dirs, patch_size=3):
        self.patient_dirs = patient_dirs
        self.patch = patch_size

        self.hist_norm = HistogramNormalize()
        self.scale = ScaleIntensityRange(0, 255, 0, 1)

        self.modalities = ['t1n', 't2f', 't2w']
        self.target = 't1c'

        # Preload all volumes into memory for speed
        self.data = []
        for p in self.patient_dirs:
            vols = []
            modalities = get_modality_paths(p)

            for m in self.modalities + [self.target]:
                # vol = nib.load(os.path.join(p, f"{m}.nii.gz")).get_fdata()
                vol = nib.load(modalities[m]).get_fdata()
                vol = vol[:,:, 15:111]
                # print(vol.shape)
                vol = torch.from_numpy(vol).float()
                vol = self.scale(self.hist_norm(vol))
                vols.append(vol)
            self.data.append(vols)

    def __len__(self):
        return 100000  # random sampling per epoch

    def __getitem__(self, idx):
        # Random patient
        vols = random.choice(self.data)

        t1n, t2f, t2w, t1c = vols

        H, W, D = t1n.shape

        # Random coordinate with safety margin
        i = random.randint(1, H-2)
        j = random.randint(1, W-2)
        k = random.randint(1, D-2)

        # Extract 3×3×3 voxel patch
        patch_in = torch.stack([
            t1n[i-1:i+2, j-1:j+2, k-1:k+2],
            t2f[i-1:i+2, j-1:j+2, k-1:k+2],
            t2w[i-1:i+2, j-1:j+2, k-1:k+2]
        ])  # shape (3,3,3,3)

        patch_out = t1c[i-1:i+2, j-1:j+2, k-1:k+2].unsqueeze(0)  # shape (1,3,3,3)

        return patch_in, patch_out
```

```python
import torch.nn as nn

class UNet3DGenerator(nn.Module):
    def __init__(self, in_channels=3, out_channels=1, features=32):
        super().__init__()

        self.enc1 = nn.Sequential(
            nn.Conv3d(in_channels, features, 3, padding=1),
            nn.ReLU(True)
        )
        self.enc2 = nn.Sequential(
            nn.Conv3d(features, features*2, 3, padding=1),
            nn.ReLU(True)
        )
        self.enc3 = nn.Sequential(
            nn.Conv3d(features*2, features*4, 3, padding=1),
            nn.ReLU(True)
        )

        self.dec2 = nn.Sequential(
            nn.Conv3d(features*4, features*2, 3, padding=1),
            nn.ReLU(True)
        )
        self.dec1 = nn.Sequential(
            nn.Conv3d(features*2, features, 3, padding=1),
            nn.ReLU(True)
        )

        self.final = nn.Conv3d(features, out_channels, 1)

    def forward(self, x):
        e1 = self.enc1(x)
        e2 = self.enc2(e1)
        e3 = self.enc3(e2)

        d2 = self.dec2(e3)
        d1 = self.dec1(d2)

        return self.final(d1)
```

```python
class PatchGAN3D(nn.Module):
    def __init__(self, in_channels=4):
        super().__init__()
        self.model = nn.Sequential(
            nn.Conv3d(in_channels, 32, 3, stride=2, padding=1), # Retains 3x3x3 kernel from previous fix
            nn.LeakyReLU(0.2),

            nn.Conv3d(32, 64, 3, stride=1, padding=1), # Changed stride from 2 to 1 to prevent collapse
            nn.InstanceNorm3d(64),
            nn.LeakyReLU(0.2),

            nn.Conv3d(64, 1, 3, stride=1, padding=1) # Ensure final layer is also consistent with stride=1
        )
```

```python
import torch.nn.functional as F
from monai.losses import SSIMLoss

adv_loss = nn.BCEWithLogitsLoss()
l1_loss = nn.L1Loss()
ssim_loss = SSIMLoss(spatial_dims=3)
```

```python
from math import log10

def psnr(pred, target):
    mse = F.mse_loss(pred, target)
    return 20 * torch.log10(1.0 / torch.sqrt(mse))

def ssim(pred, target):
    return 1 - ssim_loss(pred, target)
```

```python
def train_pix2pix(generator, discriminator, dataloader, epochs=20, lr=2e-4):
    device = "cuda" if torch.cuda.is_available() else "cpu"
    generator, discriminator = generator.to(device), discriminator.to(device)

    opt_g = torch.optim.Adam(generator.parameters(), lr=lr, betas=(0.5, 0.999))
    opt_d = torch.optim.Adam(discriminator.parameters(), lr=lr, betas=(0.5, 0.999))

    for epoch in range(epochs):
        for X, Y in dataloader:
            X, Y = X.to(device), Y.to(device)

            # =========================
            #  Train Discriminator
            # =========================
            fake = generator(X)
            real_pair = torch.cat([X, Y], dim=1)
            fake_pair = torch.cat([X, fake.detach()], dim=1)

            d_real = discriminator(real_pair)
            d_fake = discriminator(fake_pair)

            loss_d = 0.5 * (adv_loss(d_real, torch.ones_like(d_real)) +
                            adv_loss(d_fake, torch.zeros_like(d_fake)))

            opt_d.zero_grad()
            loss_d.backward()
            opt_d.step()

            # =========================
            #  Train Generator
            # =========================
            fake = generator(X)
            fake_pair = torch.cat([X, fake], dim=1)

            d_fake = discriminator(fake_pair)

            loss_g_adv = adv_loss(d_fake, torch.ones_like(d_fake))
            loss_g_l1 = l1_loss(fake, Y)
            loss_g_ssim = ssim_loss(fake, Y)

            loss_g = loss_g_adv + 100 * loss_g_l1 + 10 * loss_g_ssim

            opt_g.zero_grad()
            loss_g.backward()
            opt_g.step()

        print(f"Epoch {epoch+1}/{epochs} | "
              f"D Loss: {loss_d.item():.4f} | "
              f"G Loss: {loss_g.item():.4f} | "
              f"PSNR: {psnr(fake, Y):.2f} | "
              f"SSIM: {ssim(fake, Y):.4f}")
```
